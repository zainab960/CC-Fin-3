## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import LinearRegression
modele=LinearRegression().fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation

**Nom :** Régression linéaire (en anglais : Linear Regression).
**Famille :** régression. **Objectif :** prédire une valeur numérique à partir de variables connues.

**Étapes :** données en entrée (X et y) → le modèle calcule la droite qui passe au plus près des points → il prédit y pour de nouvelles données.

**Formule :** ŷ = β₀ + β₁x₁ + β₂x₂ + ... + βₚxₚ
- ŷ : valeur prédite
- x₁ ... xₚ : variables explicatives
- β₀ : constante (ordonnée à l'origine)
- β₁ ... βₚ : coefficients, ce sont les paramètres appris par le modèle
- Hyperparamètre : aucun dans cette version simple.

**Avantages :** facile à interpréter ; rapide à entraîner.
**Limites :** suppose une relation linéaire ; sensible aux valeurs extrêmes.

### Schéma explicatif

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

rng = np.random.default_rng(42)
x = rng.uniform(0, 10, 40).reshape(-1, 1)
y = 2 + 3 * x.ravel() + rng.normal(0, 3, 40)

m = LinearRegression().fit(x, y)
y_pred = m.predict(x)

plt.scatter(x, y, label="Observations")
plt.plot(x, y_pred, color="red", label="Droite de régression")
for xi, yi, pi in zip(x.ravel(), y, y_pred):
    plt.plot([xi, xi], [yi, pi], color="gray", linewidth=0.7)
plt.xlabel("x (variable connue)")
plt.ylabel("y (valeur à prédire)")
plt.title("Régression linéaire : droite et erreurs")
plt.legend()
plt.show()
```

**Comment lire le schéma :** les points bleus sont les observations. La droite rouge est la prédiction du modèle. Les traits gris sont les erreurs entre chaque point et la droite : le modèle choisit la droite qui rend la somme de ces erreurs au carré la plus petite.

### Contextes d'utilisation

- **Cas 1 (finance/gestion) :** prédire le chiffre d'affaires d'une entreprise selon son budget publicitaire.
- **Cas 2 :** prédire le prix d'un appartement selon sa surface.
- **Données et prérequis :** variables numériques, cible numérique, relation proche d'une droite.
- **Cas peu approprié :** prédire si un client rembourse son crédit (oui/non), car la cible n'est pas un nombre.
- **Métriques :** MAE, RMSE, R².

### Analyse de l'expérience Python

- **Entrées :** Xr_ent et yr_ent (75 % des données) pour l'apprentissage ; Xr_test et yr_test (25 %) pour le test.
- **Hyperparamètres :** aucun réglé, valeurs par défaut.
- **Sortie :** MAE = 9.35. En moyenne, la prédiction s'écarte de la vraie valeur de ce montant.
- **Ce que ça permet de conclure :** la qualité du modèle sur ce jeu de test.
- **Ce que ça ne permet pas :** les données sont synthétiques (générées par ordinateur), donc aucune conclusion sur le monde réel.

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation

**Nom :** Régression Ridge (en anglais : Ridge Regression, ou régularisation L2).
**Famille :** régression. **Objectif :** prédire une valeur numérique en limitant le surapprentissage.

**Étapes :** données en entrée (X et y) → les variables sont standardisées → le modèle apprend les coefficients en pénalisant les valeurs trop grandes → il prédit y pour de nouvelles données.

**Critère minimisé :** Σ (yᵢ − ŷᵢ)² + α Σ βⱼ²
- yᵢ : valeur réelle ; ŷᵢ : valeur prédite
- βⱼ : coefficients (paramètres appris par le modèle)
- α : force de la pénalité (hyperparamètre choisi par l'utilisateur, ici α = 1)

**Avantages :** plus stable que la régression linéaire quand les variables sont corrélées ; limite le surapprentissage.
**Limites :** ne supprime aucune variable (les coefficients se rapprochent de 0 sans l'atteindre) ; il faut choisir α et standardiser les variables.

### Schéma explicatif

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler

alphas = np.logspace(-2, 4, 50)
X_std = StandardScaler().fit_transform(Xr_ent)
coefs = [Ridge(alpha=a).fit(X_std, yr_ent).coef_ for a in alphas]

plt.plot(alphas, coefs)
plt.xscale("log")
plt.xlabel("alpha (force de la pénalité)")
plt.ylabel("Valeur des coefficients")
plt.title("Ridge : les coefficients diminuent quand alpha augmente")
plt.legend([f"Variable {i+1}" for i in range(X_std.shape[1])])
plt.show()
```

**Comment lire le schéma :** chaque courbe est le coefficient d'une variable. À gauche (alpha faible), les coefficients sont grands, comme en régression linéaire. En allant vers la droite (alpha fort), la pénalité les rapproche de zéro, sans les annuler.

### Contextes d'utilisation

- **Cas 1 (finance) :** prédire le rendement d'un portefeuille à partir de nombreux ratios financiers souvent corrélés entre eux.
- **Cas 2 (marketing) :** prédire les ventes à partir des budgets de plusieurs canaux publicitaires liés entre eux.
- **Données et prérequis :** variables numériques standardisées, cible numérique.
- **Cas peu approprié :** quand on veut éliminer les variables inutiles (le Lasso est plus adapté), ou quand la relation n'est pas linéaire.
- **Métriques :** MAE, RMSE, R².

### Analyse de l'expérience Python

- **Entrées :** Xr_ent et yr_ent (75 % des données) pour l'apprentissage ; Xr_test et yr_test (25 %) pour le test.
- **Hyperparamètres :** alpha = 1.0 (valeur fixée, non réglée) ; les variables sont standardisées avant le modèle.
- **Sortie :** MAE = 9,37, contre 9,35 pour la régression linéaire.
- **Ce que ça permet de conclure :** sur ce découpage, Ridge et la régression linéaire donnent des erreurs quasi identiques.
- **Ce que ça ne permet pas :** dire que l'un est meilleur que l'autre (écart minime, un seul découpage, données synthétiques sans variables corrélées).

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
```

### Présentation

**Nom :** Régression Lasso (en anglais : Lasso Regression, *Least Absolute Shrinkage and Selection Operator*, régularisation L1).
**Famille :** régression. **Objectif :** prédire une valeur numérique tout en sélectionnant les variables utiles.

**Étapes :** données en entrée (X et y) → les variables sont standardisées → le modèle apprend les coefficients en pénalisant leur valeur absolue, ce qui met à zéro ceux des variables peu utiles → il prédit y pour de nouvelles données.

**Critère minimisé :** (1 / 2n) Σ (yᵢ − ŷᵢ)² + α Σ |βⱼ|
- n : nombre d'observations ; yᵢ : valeur réelle ; ŷᵢ : valeur prédite
- βⱼ : coefficients (paramètres appris par le modèle)
- α : force de la pénalité (hyperparamètre, ici α = 0,1)

**Avantages :** sélectionne automatiquement les variables (certains coefficients valent exactement 0) ; modèle plus simple à interpréter.
**Limites :** instable quand plusieurs variables sont très corrélées (il en garde une au hasard) ; il faut choisir α et standardiser les variables.

### Schéma explicatif

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Lasso
from sklearn.preprocessing import StandardScaler

alphas = np.logspace(-2, 2, 50)
X_std = StandardScaler().fit_transform(Xr_ent)
coefs = [Lasso(alpha=a, max_iter=10000).fit(X_std, yr_ent).coef_ for a in alphas]

plt.plot(alphas, coefs)
plt.xscale("log")
plt.xlabel("alpha (force de la pénalité)")
plt.ylabel("Valeur des coefficients")
plt.title("Lasso : les coefficients deviennent exactement nuls")
plt.legend([f"Variable {i+1}" for i in range(X_std.shape[1])])
plt.show()
```

**Comment lire le schéma :** chaque courbe est le coefficient d'une variable. À gauche (alpha faible), les coefficients sont proches de ceux de la régression linéaire. En allant vers la droite (alpha fort), les courbes tombent à zéro et y restent : le Lasso élimine alors la variable. C'est la différence avec Ridge, où les coefficients s'approchent de zéro sans l'atteindre.

### Contextes d'utilisation

- **Cas 1 (finance) :** identifier les quelques ratios financiers qui expliquent vraiment le risque de faillite d'une entreprise parmi des dizaines de ratios.
- **Cas 2 (marketing) :** repérer les canaux publicitaires qui influencent réellement les ventes.
- **Données et prérequis :** variables numériques standardisées, cible numérique, beaucoup de variables dont certaines inutiles.
- **Cas peu approprié :** quand toutes les variables sont utiles et corrélées entre elles, car le Lasso en élimine certaines de façon arbitraire (Ridge est alors plus adapté).
- **Métriques :** MAE, RMSE, R².

### Analyse de l'expérience Python

- **Entrées :** Xr_ent et yr_ent (75 % des données) pour l'apprentissage ; Xr_test et yr_test (25 %) pour le test.
- **Hyperparamètres :** alpha = 0,1 et max_iter = 10000 (valeurs fixées, non réglées) ; les variables sont standardisées avant le modèle.
- **Sortie :** R² = 0.986 Le modèle explique environ [TON NOMBRE × 100] % de la variance de la cible sur le jeu de test.
- **Ce que ça permet de conclure :** le Lasso prédit très bien sur ce jeu de test synthétique.
- **Ce que ça ne permet pas :** le comparer à la régression linéaire et à Ridge, qui ont été évaluées avec la MAE et non le R² ; conclure sur le monde réel (données synthétiques).

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),PolynomialFeatures(degree=2,include_bias=False),Ridge(alpha=10)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation

**Nom :** Régression polynomiale (en anglais : Polynomial Regression).
**Famille :** régression supervisée. **Objectif :** prédire une valeur numérique quand la relation avec les variables n'est pas une droite.

**Intuition :** on ajoute aux données des termes de degré supérieur (x², produits x₁x₂…), puis on applique une régression linéaire sur ces nouvelles variables. Le modèle est courbe en x, mais linéaire en ses coefficients.

**Étapes :** données en entrée (X et y) → standardisation → création des termes polynomiaux → apprentissage des coefficients (ici avec la pénalité Ridge) → prédiction de y.

**Formule (une variable) :** ŷ = β₀ + β₁x + β₂x² + … + β_d x^d

**Critère minimisé (avec Ridge) :** Σ (yᵢ − ŷᵢ)² + α Σ βⱼ²
- x : variable d'entrée ; ŷᵢ : valeur prédite ; yᵢ : valeur réelle
- βⱼ : coefficients (paramètres appris par le modèle)
- d : degré du polynôme (hyperparamètre, ici d = 2)
- α : force de la pénalité (hyperparamètre, ici α = 10)

**Avantages :** simple, rapide et interprétable ; capte les courbures et les interactions entre variables.
**Limites :** le nombre de termes explose avec le degré (ici 5 variables en degré 2 donnent 20 termes), d'où un risque de surapprentissage ; instable en dehors de la plage des données d'entraînement.

### Schéma explicatif

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline

# Données simulées uniquement pour illustrer le principe
rng = np.random.RandomState(42)
x = np.sort(rng.uniform(-3, 3, 60))
y = 0.5 * x**2 - x + rng.normal(0, 0.8, 60)
X1 = x.reshape(-1, 1)
grille = np.linspace(-3, 3, 300).reshape(-1, 1)

plt.figure(figsize=(8, 5))
plt.scatter(x, y, color="gray", alpha=0.7, label="Observations (simulées pour le schéma)")
for degre, couleur, nom in [(1, "tab:red", "Degré 1 (droite)"),
                            (2, "tab:green", "Degré 2 (bon ajustement)"),
                            (10, "tab:blue", "Degré 10 (surapprentissage)")]:
    m = make_pipeline(PolynomialFeatures(degree=degre, include_bias=False),
                      StandardScaler(), LinearRegression()).fit(X1, y)
    plt.plot(grille, m.predict(grille), color=couleur, linewidth=2, label=nom)

plt.xlabel("x (variable explicative)")
plt.ylabel("y (variable à prédire)")
plt.title("Régression polynomiale : effet du degré sur l'ajustement")
plt.ylim(y.min() - 2, y.max() + 2)
plt.legend()
plt.show()
```

**Comment lire le schéma :** les points gris sont les observations (simulées ici pour illustrer). La droite rouge (degré 1) est trop rigide et ne suit pas la courbure : c'est le sous-apprentissage. La courbe verte (degré 2) suit la tendance générale. La courbe bleue (degré 10) colle aux points et oscille sur les bords : elle apprend le bruit, c'est le surapprentissage. Le degré est donc un hyperparamètre à choisir par validation, jamais sur le test final.

### Contextes d'utilisation

- **Cas 1 (marketing) :** estimer l'effet du budget publicitaire sur les ventes. Les rendements sont décroissants (chaque euro de plus rapporte moins), ce qu'une courbe capte mieux qu'une droite.
- **Cas 2 (ingénierie) :** modéliser la consommation d'un moteur selon la vitesse, avec une forme en U que le degré 2 représente bien.
- **Données et prérequis :** variables numériques standardisées, cible numérique, relation lisse ; le degré se choisit par validation croisée.
- **Cas peu approprié :** prédire loin en dehors de la plage observée (par exemple des ventes dans plusieurs années), car un polynôme diverge vite hors des données. Il est aussi peu adapté avec beaucoup de variables ou des ruptures brusques (une forêt aléatoire convient mieux).
- **Métriques :** MAE, RMSE, R², toujours comparées à celles de la régression linéaire simple.

### Analyse de l'expérience Python

## 5. Arbre de régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

PrésentationNom : Arbre de décision pour la régression (en anglais : Decision Tree Regressor).Famille : régression supervisée non paramétrique.Objectif : prédire une valeur numérique en découpant l'espace des variables en régions par des règles logiques successives (du type "Si $X_1 > a$ et $X_2 \le b$").Intuition : l'arbre pose une série de questions binaires sur les caractéristiques. Chaque question divise les données en sous-groupes plus homogènes. Pour prédire la valeur d'une nouvelle observation, on suit le chemin correspondant et on lui attribue la moyenne de la valeur cible des observations d'entraînement situées dans la même feuille finale.Étapes : données en entrée ($X$ et $y$) → recherche du meilleur découpage (variable et seuil) à chaque nœud pour réduire la variance → répulsion/division jusqu'à atteindre un critère d'arrêt (profondeur max, nombre min d'échantillons par feuille) → prédiction par la moyenne de la feuille.Formule (prédiction dans une feuille $R_m$) : $\hat{y} = \frac{1}{N_m} \sum_{i \in R_m} y_i$Critère de séparation minimisé (Variance / MSE) :$\text{MSE}_{nœud} = \frac{1}{N} \sum_{i=1}^{N} (y_i - \bar{y})^2$$N_m$ : nombre d'observations dans la feuille ou la région $R_m$$y_i$ : valeur réelle de la cible pour l'observation $i$$\bar{y}$ : moyenne des valeurs cibles dans la région considéréeAvantages : non paramétrique (ne suppose aucune forme linéaire) ; capte naturellement les interactions complexes et non linéaires ; très facile à interpréter et à visualiser si l'arbre n'est pas trop profond ; insensible aux échelles des variables (pas besoin de standardisation).Limites : forte tendance au surapprentissage (overfitting) si la profondeur n'est pas limitée ; modèle instable (une petite modification des données peut changer complètement l'arbre) ; prédictions constantes par morceaux (pas de lissage/extrapolation hors des bornes d'entraînement).

### Schéma explicatif

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor

# Données simulées pour illustrer le principe
rng = np.random.RandomState(42)
X_sim = np.sort(rng.uniform(-3, 3, 80)).reshape(-1, 1)
y_sim = np.sin(X_sim).ravel() + rng.normal(0, 0.1, 80)

grille = np.linspace(-3, 3, 500).reshape(-1, 1)

plt.figure(figsize=(10, 5))
plt.scatter(X_sim, y_sim, color="gray", alpha=0.7, label="Observations (simulées)")

for prof, couleur, nom in [(2, "tab:green", "Profondeur 2 (sous-ajustement)"),
                           (5, "tab:blue", "Profondeur 5 (bon ajustement)"),
                           (15, "tab:red", "Profondeur 15 (surapprentissage)")]:
    m = DecisionTreeRegressor(max_depth=prof, random_state=42).fit(X_sim, y_sim)
    plt.plot(grille, m.predict(grille), color=couleur, linewidth=2, label=nom)

plt.xlabel("x (variable explicative)")
plt.ylabel("y (variable à prédire)")
plt.title("Arbre de décision : effet de la profondeur max (max_depth) sur les prédictions")
plt.legend()
plt.show()
```

Comment lire le schéma : les points gris représentent les observations. La courbe en escalier montre la nature des prédictions d'un arbre : il prédit une valeur constante par palier (la moyenne du sous-groupe).

Avec max_depth = 2 (ligne verte), l'arbre est trop simple et ne réalise que 4 paliers : c'est le sous-apprentissage.

Avec max_depth = 5 (ligne bleue), l'arbre capture bien la forme sinusoïdale de la fonction.

Avec max_depth = 15 (ligne rouge), l'arbre crée des paliers minuscules pour ajuster chaque point individuel, y compris le bruit : c'est le surapprentissage. La profondeur maximale (max_depth) est donc un hyperparamètre critique à réguler.

Contextes d'utilisationCas 1 (Immobilier) : estimer le prix d'un logement en fonction de critères discrets ou continus (nombre de pièces, présence de balcon, quartier), où les relations entre variables sont souvent par seuils.Cas 2 (Tarification d'assurance) : segmenter les assurés en groupes de risque homogènes pour calculer une prime moyenne par segment.Données et prérequis : fonctionne sur des données numériques et catégorielles (après encodage) ; ne nécessite pas de normalisation des variables ; tolère bien les relations non linéaires.Cas peu approprié : extrapolation temporelle (incapable de prédire des tendances à la hausse hors de la plage d'apprentissage) ; relations parfaitement linéaires simples (où une régression linéaire fera mieux avec moins de complexité).Métriques : MAE, RMSE, $R^2$.

### Présentation

**Nom :** Régression logistique (en anglais : Logistic Regression).
**Famille :** Classification supervisée (utilisée pour prédire une probabilité ou une classe binaire, ex: 0 ou 1, bon ou mauvais payeur).
**Objectif :** Modéliser la relation entre une variable cible qualitative et un ensemble de variables explicatives à l'aide d'une fonction sigmoïde.

### Formulation mathématique

La probabilité $P(Y=1|X)$ est modélisée par la fonction logistique (sigmoïde) :
$$P(Y=1|X) = \frac{1}{1 + e^{-(\beta_0 + \beta_1 X_1 + \beta_2 X_2 + \dots + \beta_n X_n)}}$$

Où :
- $\beta_0$ est l'ordonnée à l'origine.
- $\beta_1, \beta_2, \dots$ sont les coefficients associés à chaque variable financière/explicative.

### Avantages et Limites

* **Avantages :** Simple à interpréter, très rapide à entraîner, donne des probabilités claires pour l'aide à la décision (ex: scoring de crédit).
* **Limites :** Incapable de capturer des relations non-linéaires complexes sans transformation préalable des variables.

### Contexte d'utilisation en finance

* **Scoring de crédit :** Estimer la probabilité qu'un client fasse défaut sur son crédit bancaire.
* **Détection de faillite :** Prédire si une entreprise risque de faire faillite en fonction de ses ratios financiers.

### Présentation

**Nom :** Machines à Vecteurs de Support (en anglais : Support Vector Machines - SVM).
**Famille :** Apprentissage supervisé (classification et régression).
**Objectif :** Trouver l'hyperplan optimal qui sépare au mieux les différentes classes dans un espace à $n$ dimensions en maximisant la marge entre ces classes.

### Formulation mathématique

Le modèle cherche à maximiser la marge $\frac{2}{\|\mathbf{w}\|}$ tout en satisfaisant les contraintes de classification :
$$y_i (\mathbf{w} \cdot \mathbf{x}_i + b) \ge 1, \quad \forall i$$

Où :
- $\mathbf{w}$ est le vecteur normal à l'hyperplan.
- $b$ est le biais.
- $y_i$ est la classe associée (-1 ou 1).
*(En cas de non-linéarité, une fonction de "noyau" (Kernel, ex: RBF) est utilisée pour projeter les données dans un espace de dimension supérieure).*

### Avantages et Limites

* **Avantages :** Très efficace en grande dimension, robuste face au surapprentissage (grâce à la maximisation de la marge), et puissant pour modéliser des frontières complexes via les noyaux.
* **Limites :** Sensible aux valeurs aberrantes (*outliers*), temps de calcul élevé sur les très grands volumes de données, et difficulté d'interprétation directe des coefficients.

### Contexte d'utilisation en finance

* **Scoring de crédit :** Séparer de manière optimale les clients solvables des clients à haut risque de défaut.
* **Prédiction de tendance boursière :** Classifier les signaux de marché (achat / vente) en fonction des indicateurs techniques.

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import pandas as pd

# Instanciation et entraînement du modèle SVM (avec un noyau RBF par défaut)
svm_model = SVC(kernel='rbf', C=1.0, random_state=42)
svm_model.fit(X_train, y_train)

# Prédiction sur l'ensemble de test
y_pred = svm_model.predict(X_test)

# Affichage des résultats
print("Précision (Accuracy) :", accuracy_score(y_test, y_pred))
print("\nRapport de classification :\n", classification_report(y_test, y_pred))
```

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```
